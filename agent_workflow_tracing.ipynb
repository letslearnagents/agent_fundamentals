```python
instructions = """
You are a helpful market research agent who will perform the requeired calcculation to get the result

"""
```

```python
@function_tool
def get_speed_of_light():
    """
    Returns the speed of light in metres per second.
    """
    return 299_792_458.0
```

```python
@function_tool
def calculate_square_root(number : float) -> float:
    """
    calculate the square root of a given number
    """
    return math.sqrt(number)
```

```python
agent = Agent(
name = "math square root calculator tool",
instructions = instructions,
model = 'gpt-4o-mini',
tools = [get_speed_of_light, calculate_square_root]
)
```

```python
response = await Runner.run(
    starting_agent = agent,
    input = "Calculate the square root of the speed of light and return the final result."
)
```

```python
response.final_output
```

```python
async def run_with_full_trace():
    print("\n" + "="*60)
    print("STARTING AGENT RUN")
    print("="*60)

    result = await Runner.run(
        agent,
        "Calculate the square root of the speed of light and return the final result."
    )

    print("\n" + "="*60)
    print("FULL TRACE — EVERY STEP")
    print("="*60)

    for i, item in enumerate(result.new_items):
        print(f"\n--- Item {i+1}: {type(item).__name__} ---")
        print(item)

    print("\n" + "="*60)
    print("FINAL ANSWER")
    print("="*60)
    print(result.final_output)

# asyncio.run(run_with_full_trace())
await run_with_full_trace()
```
