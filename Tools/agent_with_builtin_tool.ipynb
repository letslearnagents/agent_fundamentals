## Define the agent

```python
agent = Agent(
    name = "web search tool",
    instructions = "You are a helpful agent. search the web to find the best response for user's query. If some response is not available, feel free to clearly send the message about it,",
    model = "gpt-4o-mini",
    tools = [WebSearchTool()]
)
```

```python
agent
```

## Response 

```python
response = await Runner.run(
    starting_agent = agent,
    input = "What are the top Python frameworks for building AI Agents? Compare their features, advantages, and limitations"
)
    
```

## Print the raw response

```python
print(response.final_output)
```

## View the tool calling traces

```python
for item in response.new_items:
    print(item)
```
