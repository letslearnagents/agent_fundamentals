## Define the instructions for our agent

```python
agent_instructions = """
You are an English dictionary assistant.

Your task is to verify whether the user's understanding of a word is correct.

Input Format:
word : meaning

Instructions:
- Analyze the meaning using your own knowledge.
- Do not assume the user is correct.
- Validate whether the meaning is accurate or not.
- If the meaning is partially correct, mention that briefly.
- Keep the response concise.

Output Format:
True/False
Short one-line explanation.
Synonym: <synonym>
"""
```

## Define the agent

```python
agent = Agent(
name = "English Dictionary Agent",
instructions = agent_instructions,
model = "gpt-4o-mini"   
    
)
```

```python
print(f"Our first agent {agent.name} is created.")
```

## Provide the first input to our agent

```python
stmt1 = "wonderful : Something which is beautiful"
```

## Get the response from the model

```python
response = await Runner.run(
starting_agent = agent,
input = stmt1    
)
```

## Print the final response

```python
print(response.final_output)
```
