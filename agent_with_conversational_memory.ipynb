## Define the instructions for the agent

```python
instructions = """
You are a helpful travel assistant who will help with my stay in one of the tourist properties

"""
```

## Define an agent

```python
agent = Agent(
    name = "travel assistant agent",
    instructions = instructions,
    model = "gpt-4o-mini" ,
    tools = []
)
```

## Define the query

```python
query1 = """

My name is Jonny. I am a travel enthusiast.
I love travelling in mountains and always prefer to stay close to nature.
I have visited many places in the past and my experience has been amazing except a place where the place was close to nature but because of being a popular tourist place it was flooded by visiotrs and due to which the stay was all filled with honking and pollution.
Suggest some places near Himachal where I can plan my 5 days stay.

"""
```

```python
query2 = {
"role" : "user",
"content": "What is my name"
}
```

```python
query3 = {
"role" : "user",
"content": "What are my preferences when it comes to travel?"
}
```

## Define the variable to store the chat history

```python
history = []
```

## Call the agent with a query

```python
response1 = await Runner.run(
starting_agent = agent,
input = query1
    
)
```

## Print the response for the first query

```python
response1.final_output
```

```python
response1.to_input_list()
```

## Store this response which we see above as a list into the "history" variable that we have defined earlier

```python
history = response1.to_input_list()
```
